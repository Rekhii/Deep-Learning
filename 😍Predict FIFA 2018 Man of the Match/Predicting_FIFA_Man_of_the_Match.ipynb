##Preprocessing

```python
df = df.drop(columns=["Date"])
```

```python
df = df.drop(columns=["Own goals"])
df = df.drop(columns=["Own goal Time"])
```

```python
df["1st Goal"] = df["1st Goal"].fillna(df["1st Goal"].mean())
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df["Round"].unique()
```

```python
round_values = list(df['Round'].unique())
round_values
```

```python
round_mappings = {label: index for index, label in enumerate(round_values)}
round_mappings
```

```python
df['Round'] = df['Round'].apply(lambda x: round_mappings[x])
```

```python
df["PSO"].value_counts()
```

```python
df["PSO"] = df["PSO"].replace({"No": 0 , "Yes": 1})
```

```python
df["Man of the Match"] = df["Man of the Match"].replace({"No": 0 , "Yes": 1})
```

```python
df['Team'].unique()
```

```python
df['Opponent'].unique()
```

```python
pd.get_dummies(df['Opponent'].apply(lambda x: "opp_" + x))
```

```python
df['Opponent'] = df['Opponent'].apply(lambda x: "opp_" + x)
```

```python
data_concat = pd.concat([df, pd.get_dummies(df['Team']), pd.get_dummies(df['Opponent'])], axis=1)
```

```python
data_concat.drop(['Team', 'Opponent'], axis=1, inplace=True)
```

```python
np.sum(data_concat.dtypes == 'object')
```

```python
data_concat
```

```python
data_concat.info()
```

```python
# Convert boolean columns to integers (0s and 1s)
for col in data_concat.columns:
  if data_concat[col].dtype == 'bool':
    data_concat[col] = data_concat[col].astype(int)

display(data_concat.head())
```

##Training and Building the model


```python
y = data_concat['Man of the Match']
X = data_concat.drop('Man of the Match', axis=1)
```

```python
scaler = RobustScaler()  # work good on outliear copair to minmax and standerdscaler

X = pd.DataFrame(scaler.fit_transform(X), columns=X.columns)
```

```python
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7)
```

```python
# ML algo
sk_model = MLPClassifier(hidden_layer_sizes=(16, 16))
sk_model.fit(X_train, y_train)
```

```python
# nns with dropout
inputs = tf.keras.Input(shape=(X_train.shape[1],))
x = tf.keras.layers.Dense(128, activation=tf.nn.relu)(inputs)
x = tf.keras.layers.Dropout(0.3)(x)
x = tf.keras.layers.Dense(128, activation=tf.nn.relu)(x)
x = tf.keras.layers.Dropout(0.3)(x)
outputs = tf.keras.layers.Dense(2, activation=tf.nn.softmax)(x)

tf_model = tf.keras.Model(inputs=inputs, outputs=outputs)
```

```python
tf_model.compile(
    optimizer='adam',
    loss=tf.keras.losses.SparseCategoricalCrossentropy(),
    metrics=['accuracy']
)
```

```python
tf_model.fit(
    X_train,
    y_train,
    validation_split=0.2,
    batch_size=32,
    epochs=224
)
```

```python
sk_score = sk_model.score(X_test, y_test)
tf_score = tf_model.evaluate(X_test, y_test, verbose=False)
```

```python
print(f"ML Model: {sk_score}")
print(f"NNs: {tf_score[1]}")
```
